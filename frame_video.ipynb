```python

import cv2
import os
import glob

# ---- Settings ----
input_folder = "more_frames"
output_folder = "frames"
start_time_sec = 1
frames_per_video = 10
# ------------------

os.makedirs(output_folder, exist_ok=True)

# Get all .mp4 files in the input folder
video_paths = glob.glob(os.path.join(input_folder, "*.mp4"))

if not video_paths:
    print(f"No .mp4 files found in '{input_folder}'")

for video_path in sorted(video_paths):
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"Could not open {video_path}")
        continue

    fps = cap.get(cv2.CAP_PROP_FPS)

    if fps <= 0:
        print(f"Invalid FPS for {video_path}")
        cap.release()
        continue

    start_frame = int(start_time_sec * fps)
    cap.set(cv2.CAP_PROP_POS_FRAMES, start_frame)

    base_name = os.path.splitext(os.path.basename(video_path))[0]

    saved = 0

    while saved < frames_per_video:
        ret, frame = cap.read()

        if not ret:
            print(f"{video_path}: video ended after {saved} frames")
            break

        output_path = os.path.join(
            output_folder,
            f"{base_name}_frame_{saved:02d}.jpg"
        )

        cv2.imwrite(output_path, frame)
        saved += 1

    cap.release()
    print(f"{video_path}: saved {saved} frames")

print("Done!")
```
